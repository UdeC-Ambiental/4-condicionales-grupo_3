# 🧩 Reto 4 – Integración de resultados
**Objetivo:** combinar resultados parciales en un diagnóstico integral por sitio.

### Pasos
1. Carga `resultado_temperatura.csv`, `resultado_ph.csv`, `resultado_oxigeno.csv`.
2. Únelos por `sitio` usando `merge`.
3. Crea `diagnostico_final` con reglas (parcial/total/normal).
4. Exporta `diagnostico_final.csv`.
5. Cuenta categorías con `value_counts()`.

> **Reto adicional:** gráfico de barras de distribución.

```python
import pandas as pd  # Importamos la librería pandas para trabajar con datos tipo tabla

# 1️⃣ Cargar los tres archivos CSV
temperatura = pd.read_csv('resultado_temperatura.csv')
ph = pd.read_csv('resultado_ph.csv')
oxigeno = pd.read_csv('resultado_oxigeno.csv')

# 2️⃣ Unirlos por la columna "sitio" usando merge
# Esto junta los datos de cada archivo según el mismo nombre de sitio.
df = temperatura.merge(ph, on='sitio').merge(oxigeno, on='sitio')

# 3️⃣ Crear una nueva columna con el diagnóstico final
# Usamos condicionales (if, elif, else) dentro de una función.

def diagnostico(row):
    # Si alguno de los resultados es "malo", el diagnóstico es parcial.
    if (row['resultado_temperatura'] == 'malo') or (row['resultado_ph'] == 'malo') or (row['resultado_oxigeno'] == 'malo'):
        return 'parcial'
    
    # Si todos los resultados son "buenos", el diagnóstico es total.
    elif (row['resultado_temperatura'] == 'bueno') and (row['resultado_ph'] == 'bueno') and (row['resultado_oxigeno'] == 'bueno'):
        return 'total'
    
    # Si no cumple las anteriores (por ejemplo mezcla de bueno y normal), será normal.
    else:
        return 'normal'

# Aplicamos la función a cada fila del DataFrame
df['diagnostico_final'] = df.apply(diagnostico, axis=1)

# 4️⃣ Exportar el resultado a un nuevo archivo CSV
df.to_csv('diagnostico_final.csv', index=False)

# 5️⃣ Contar cuántos sitios quedaron en cada categoría
conteo = df['diagnostico_final'].value_counts()
print(conteo)

# 6️⃣ (Reto adicional) Crear gráfico de barras
conteo.plot(kind='bar', title='Distribución del diagnóstico final')

```
