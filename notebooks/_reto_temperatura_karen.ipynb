# 🧩 Reto 1 – Temperatura
**Objetivo:** clasificar la temperatura del agua y evaluar posible estrés térmico.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `temperatura` en `df_temp`.
3. Crea `categoria_temp` con np.where o .loc.
4. Crea `riesgo_termico`.
5. Filtra a `df_caliente` (alta temperatura).
6. Exporta `resultado_temperatura.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo térmico.

```python
import pandas as pd

# 1. Cargar el archivo
df = pd.read_csv('../data/muestras_calidad_agua.csv')

# 2. Seleccionar columnas necesarias
df_temp = df[['sitio', 'temperatura']].copy()

# 3. Crear listas vacías
clasificacion = []
alerta = []

# 4. Recorrer cada temperatura con un bucle for
for valor in df_temp['temperatura']:
    if valor < 15:
        clasificacion.append('Fría')
        alerta.append('Sí')
    elif valor > 25:
        clasificacion.append('Caliente')
        alerta.append('Sí')
    else:
        clasificacion.append('Templada')
        alerta.append('No')

# 5. Agregar las listas como nuevas columnas al DataFrame
df_temp['clasificacion_temp'] = clasificacion
df_temp['alerta_termica'] = alerta

# 6. Mostrar el DataFrame con resultados
print("✅ Clasificación de temperatura con condicionales:\n")
display(df_temp)

# 7. Filtrar solo las filas con alerta térmica
df_alerta = df_temp[df_temp['alerta_termica'] == 'Sí']

# 8. Calcular estadísticas básicas
promedio = df_temp['temperatura'].mean()
desv_std = df_temp['temperatura'].std()

print(f"\n📊 Promedio de temperatura: {promedio:.2f} °C")
print(f"📉 Desviación estándar: {desv_std:.2f} °C")

# 9. Guardar el resultado en un nuevo archivo
df_alerta.to_csv('../data/resultado_temperatura.csv', index=False)
print("\n💾 Archivo 'resultado_temperatura.csv' guardado en la carpeta 'data'.")
```
